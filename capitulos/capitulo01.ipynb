```python
def nussinov(sequence):
    n = len(sequence)
    dp = [[0] * n for _ in range(n)]

    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1
            # Caso 1: i e j nao pareiam
            dp[i][j] = dp[i+1][j]

            # Caso 2: i e j pareiam
            for k in range(i, j):
                if can_pair(sequence[i], sequence[k]):
                    score = dp[i+1][k-1] + 1 + dp[k+1][j]
                    dp[i][j] = max(dp[i][j], score)

    return dp[0][n-1]

    def can_pair(base1, base2):
        """
     Verifica se duas bases de RNA podem formar um par válido.
     Pares padrão de Watson-Crick: A-U e C-G.
     """
        pares_validos = {('A', 'U'), ('U', 'A'), ('C', 'G'), ('G', 'C')}
    return (base1, base2) in pares_validos
```
