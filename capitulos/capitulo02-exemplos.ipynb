## Filtragem de Sequências FASTA

**Requisição para a IA:** *"Escreva um script em Python usando Biopython que leia `sequencias.fasta` e salve em `filtrados.fasta` apenas as sequências com mais de 500 pares de base."*

O código gerado abaixo resolve exatamente esse problema. Lembre-se que você precisará criar ou baixar um arquivo `sequencias.fasta` para que ele funcione na prática.

```python
from Bio import SeqIO

records = []
try:
    # Lê as sequências do arquivo de entrada
    for record in SeqIO.parse("sequencias.fasta", "fasta"):
        if len(record.seq) > 500:
            records.append(record)
    
    # Salva as filtradas no arquivo de saída
    SeqIO.write(records, "filtrados.fasta", "fasta")
    print(f"Filtragem concluída! Foram mantidas {len(records)} sequências.")
except FileNotFoundError:
    print("Erro: Arquivo 'sequencias.fasta' não encontrado no diretório atual. Crie este arquivo para testar o script.")
```
